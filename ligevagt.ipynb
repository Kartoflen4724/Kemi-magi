```python
import sympy as sp
from sympy import *
xi=sp.symbols("xi")

#Input substances data:
for i in range(1):
    kp=56
    p=1
    I1, nI1, cI1 = "CH4", 10, -2
    I2, nI2, cI2 = "", 0, 0
    I3, nI3, cI3 = "", 0, 0
    O1, nO1, cO1 = "", 0, 0
    O2, nO2, cO2 = "C2H2", 0, 1
    O3, nO3, cO3 = "H2", 0, 3
    inert, n_inert = "N2", 0

#n(eq) & Xi-brøker
for i in range(1):
    nIeq1=nI1+cI1*xi
    nIeq2=nI2+cI2*xi
    nIeq3=nI3+cI3*xi
    nOeq1=nO1+cO1*xi
    nOeq2=nO2+cO2*xi
    nOeq3=nO3+cO3*xi
    neqtotal=nIeq1+nIeq2+nIeq3+nOeq1+nOeq2+nOeq3+n_inert
#    print(neqtotal)
    xiIeq1=nIeq1/neqtotal
    xiIeq2=nIeq2/neqtotal
    xiIeq3=nIeq3/neqtotal
    xiOeq1=nOeq1/neqtotal
    xiOeq2=nOeq2/neqtotal
    xiOeq3=nOeq3/neqtotal
    xi_inert=n_inert/neqtotal

#xi og alpha
for i in range(1):
    #find øvre grænse for xi (den begrænsende reaktant)
    limits = []
    for n, c in [(nI1, cI1), (nI2, cI2), (nI3, cI3)]:
        if c < 0:
            limits.append(n / abs(c))
    xi_max = min(limits)
    #ky
    ky=((xiOeq1**cO1)*(xiOeq2**cO2)*(xiOeq3**cO3))/(
    (xiIeq1**abs(cI1))*(xiIeq2**abs(cI2))*(xiIeq3**abs(cI3))
    )
    delta_n=(cO1+cO2+cO3)+(cI1+cI2+cI3)
    #find xi
    lig=Eq(ky*(p**delta_n),kp)
    xi_solution=sp.solve(lig,xi)

    for sol in xi_solution:
        val = sol.evalf()
        if val.is_real and 0 < val < xi_max:
            xii = val
            break

    alpha=(abs(cI1)*xii)/nI1

    print(f"\u03be = {xii:.4f}")
    print(f"\u03b1 = {alpha:.4f}")

#Stofmængder og molbrøker
for i in range (1):
    n_I1=nIeq1.subs(xi, xii)
    n_I2=nIeq2.subs(xi, xii)
    n_I3=nIeq3.subs(xi, xii)
    n_O1=nOeq1.subs(xi, xii)
    n_O2=nOeq2.subs(xi, xii)
    n_O3=nOeq3.subs(xi, xii)

    y_I1=xiIeq1.subs(xi, xii)
    y_I2=xiIeq2.subs(xi, xii)
    y_I3=xiIeq3.subs(xi, xii)
    y_O1=xiOeq1.subs(xi, xii)
    y_O2=xiOeq2.subs(xi, xii)
    y_O3=xiOeq3.subs(xi, xii)

    subs = [
        (I1, n_I1, y_I1), (I2, n_I2, y_I2), (I3, n_I3, y_I3),
        (O1, n_O1, y_O1), (O2, n_O2, y_O2), (O3, n_O3, y_O3)]

    if n_inert > 0:
        subs.append((inert, n_inert, xi_inert.subs(xi, xii)))

    print("Ligevægtssammensætning:")
    print(f"{'Stof':<6} | {'Stofmængde':>12} | {'Molbrøk':<3}  {'|   Partialtryk':>24}")

    n_sum, y_sum, p_sum = 0, 0, 0

    for sub, n_val, y_val in subs:
        if sub != "":
            p_val = y_val * p
            n_sum += n_val
            y_sum += y_val
            p_sum += p_val
            print(f"{sub:<6} | {n_val:>8.4f} mol | {y_val:>6.4f} ({y_val*100:>6.2f} %) | {p_val:>8.4f} bar")
    print(f"{'Total':<6} | {n_sum:>8.4f} mol | {y_sum:>6.4f} ({y_sum*100:>6.2f} %) | {p_sum:>8.4f} bar")
```
