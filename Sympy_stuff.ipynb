```python
# https://docs.sympy.org/latest/index.html

# Import stuff
import sympy as sp
# Import all:
from sympy import *

import matplotlib.pyplot as plt
import numpy as np
```

```python
# Do the solve for numbers:

x, y, z = symbols('x y z')

eq1 = x**2 + 1
eq2 = y**2 + 1
eq3 = z**2 + 1

nonlinsolve([eq1, eq2, eq3], [x, y, z])


# Simplify an expression:

simplify(sin(x)**2 + cos(x)**2)

# Expand an expression:
expand((x + 1)**2)

# Factor the expression:
factor(x**3 - x**2 + x - 1)

```

```python
# Triangle stuff
# asin= inverse of sin

asin(1)
```
