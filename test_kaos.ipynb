```python
import sympy as sp
from sympy import *
import numpy as np
from numpy import *
xi=sp.symbols("xi")

#Input substances data:
for i in range(1):
    kp=56
    p=1
    nI1=10
    cI1=-2
    nI2=0
    cI2=0
    nI3=0
    cI3=0
    nO1=0
    cO1=0
    nO2=0
    cO2=1
    nO3=0
    cO3=3

#n(eq)
for i in range(1):
    nIeq1=nI1+cI1*xi
    nIeq2=nI2+cI2*xi
    nIeq3=nI3+cI3*xi
    nOeq1=nO1+cO1*xi
    nOeq2=nO2+cO2*xi
    nOeq3=nO3+cO3*xi
    neqtotal=nIeq1+nIeq2+nIeq3+nOeq1+nOeq2+nOeq3
    print(neqtotal)

#Xi-brök
for i in range(1):
    xiIeq1=nIeq1/neqtotal
    xiIeq2=nIeq2/neqtotal
    xiIeq3=nIeq3/neqtotal
    xiOeq1=nOeq1/neqtotal
    xiOeq2=nOeq2/neqtotal
    xiOeq3=nOeq3/neqtotal

#ky
ky=((xiOeq1**cO1)+(xiOeq2**cO2)+(xiOeq3**cO3))/((xiIeq1**cI1)+(xiIeq2**cI2)+(xiIeq3**cI3))
#find xi
lig=Eq(ky,kp*p)
xi_solution=sp.solve(lig,xi)
for el in xi_solution:
    print(el.evalf())
lig.simplify


```
