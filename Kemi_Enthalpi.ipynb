```python
# http://pyromat.org/doc_intro.html

# Import stuff
import pyromat as pm
import sympy as sp
from sympy import *
import matplotlib.pyplot as plt
import numpy as np

# See config settings:
#pm.config

# Skift enheden fra 'kg' til 'kmol'
pm.config['unit_matter'] = 'kmol'
```

```python
# Define the substances as gases
N2 = pm.get('ig.N2')
NO = pm.get('ig.NO')
O2 = pm.get('ig.O2')
C = pm.get('ig.C')
CO = pm.get('ig.CO')
CO2 = pm.get('ig.CO2')
H2O = pm.get('ig.H2O')
H2 = pm.get('ig.H2')
CH4 = pm.get('ig.CH4')
C2H6 = pm.get('ig.C2H6')
C3H8 = pm.get('ig.C3H8')
C4H10 = pm.get('ig.C4H10')
```

```python
#Define the substances as liquids
H2O_g = pm.get('ig.H2O')
H2O_f = pm.get('mp.H2O')
H2O_fg = H2O_f.h(T=298.15, x=1)-H2O_f.h(T=298.15, x=0)
H2O_l = H2O_g.h(T=298.15) - H2O_fg

#Calculate the enthalpy of a substance in liquid state at a given temperature.
def make_liquid(substance, Temp):
    
    substance_g = pm.get(f'ig.{substance}')
    substance_f = pm.get(f'mp.{substance}')
    substance_fg = substance_f.h(T=Temp, x=1) - substance_f.h(T=Temp, x=0)
    substance_l = substance_g.h(T=Temp) - substance_fg
    return substance_l[0]
```

```python
#Calc a reaction enthalpy for gasses
#If you dont have 3 substances going in/out, let it stay H2O with count 0
# Use H2O_l if you have liquid water

# Set you equation:
for i in range(1):
    temp_in = 273.15+227
    temp_out = 273.15+227
    input_substance = "C2H4"
    input_substance_count = 200
    input_substance2 = "O2"
    input_substance2_count = 200
    input_substance3 = "C2H4O"
    input_substance3_count = 0
    input_substance4 = "H2O"
    input_substance4_count = 0
    input_substance5 = "CO2"
    input_substance5_count = 80
    input_substance6 = "N2"
    input_substance6_count = 1520
    output_substance = "C2H4"
    output_substance_count = 160
    output_substance2 = "O2"
    output_substance2_count = 130
    output_substance3 = "C2H4O"
    output_substance3_count = 20
    output_substance4 = "H2O"
    output_substance4_count = 40
    output_substance5 = "CO2"
    output_substance5_count = 120
    output_substance6 = "N2"
    output_substance6_count = 1520


# Shorten the names to make equation eaiser to edit
for i in range(1):
    is1=input_substance
    isc1=input_substance_count 
    is2=input_substance2 
    isc2=input_substance2_count 
    is3=input_substance3 
    isc3=input_substance3_count 
    is4=input_substance4
    isc4=input_substance4_count 
    is5=input_substance5 
    isc5=input_substance5_count 
    is6=input_substance6
    isc6=input_substance6_count    
    os1=output_substance 
    osc1=output_substance_count 
    os2=output_substance2 
    osc2=output_substance2_count
    os3=output_substance3 
    osc3=output_substance3_count 
    os4=output_substance4 
    osc4=output_substance4_count 
    os5=output_substance5 
    osc5=output_substance5_count
    os6=output_substance6 
    osc6=output_substance6_count 

# Get the values for the substances in
def gI(sub):
    if sub == "H2O_l":
        substance_enthalpy=H2O_l
    else:    
        substance = pm.get(f'ig.{sub}')
        substance_enthalpy = substance.h(T=temp_in)
    return substance_enthalpy

# Get the values for the substances out
def gO(sub):
    if sub == "H2O_l":
        substance_enthalpy=H2O_l
    else:    
        substance = pm.get(f'ig.{sub}')
        substance_enthalpy = substance.h(T=temp_out)
    return substance_enthalpy

# Doing the calculation
H_out=(osc1*gO(os1))+(osc2*gO(os2))+(osc3*gO(os3))+(osc4*gO(os4))+(osc5*gO(os5))+(osc6*gO(os6))
H_in=(isc1*gI(is1))+(isc2*gI(is2))+(isc3*gI(is3))+(isc4*gI(is4))+(isc5*gI(is5))+(isc6*gI(is6))
H_total=(H_out-H_in)
print(f'{H_in[0]/1000}' " KJ/mol, in")
print(f'{H_out[0]/1000}' " KJ/mol, out")
print(f'{H_total[0]/1000}' " KJ/mol H_out-H_in")
print(f'{H_total[0]}'" kJ/h")
print(f'{H_total[0]/3600}'" kW")
```
