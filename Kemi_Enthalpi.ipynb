```python
# http://pyromat.org/doc_intro.html

# Import stuff
import sympy as sp
import pyromat as pm
import matplotlib.pyplot as plt
import numpy as np

# Skift enheden fra 'kg' til 'mol'
pm.config['unit_matter'] = 'mol'
```

```python
# Define the substances as gases
N2 = pm.get('ig.N2')
NO = pm.get('ig.NO')
O2 = pm.get('ig.O2')
C = pm.get('ig.C')
CO = pm.get('ig.CO')
CO2 = pm.get('ig.CO2')
H2O = pm.get('ig.H2O')
H2 = pm.get('ig.H2')
CH4 = pm.get('ig.CH4')
C2H6 = pm.get('ig.C2H6')
C3H8 = pm.get('ig.C3H8')
C4H10 = pm.get('ig.C4H10')
```

```python
#Define the substances as liquids
H2O_g = pm.get('ig.H2O')
H2O_f = pm.get('mp.H2O')
H2O_fg = H2O_f.h(T=298.15, x=1)-H2O_f.h(T=298.15, x=0)
H2O_l = H2O_g.h(T=298.15) - H2O_fg

#Calculate the enthalpy of a substance in liquid state at a given temperature.
def make_liquid(substance, Temp):
    
    substance_g = pm.get(f'ig.{substance}')
    substance_f = pm.get(f'mp.{substance}')
    substance_fg = substance_f.h(T=Temp, x=1) - substance_f.h(T=Temp, x=0)
    substance_l = substance_g.h(T=Temp) - substance_fg
    return substance_l[0]
```
