```python
# http://pyromat.org/doc_intro.html

# Import stuff
import sympy as sp
from sympy import *
import pyromat as pm
import matplotlib.pyplot as plt
import numpy as np

# See config settings:
#pm.config

# Skift enheden fra 'kg' til 'kmol'
pm.config['unit_matter'] = 'kmol'
```

```python
# Define the substances as gases
N2 = pm.get('ig.N2')
NO = pm.get('ig.NO')
O2 = pm.get('ig.O2')
C = pm.get('ig.C')
CO = pm.get('ig.CO')
CO2 = pm.get('ig.CO2')
H2O = pm.get('ig.H2O')
H2 = pm.get('ig.H2')
CH4 = pm.get('ig.CH4')
C2H6 = pm.get('ig.C2H6')
C3H8 = pm.get('ig.C3H8')
C4H10 = pm.get('ig.C4H10')
```

```python
#Define the substances as liquids
H2O_g = pm.get('ig.H2O')
H2O_f = pm.get('mp.H2O')
H2O_fg = H2O_f.h(T=298.15, x=1)-H2O_f.h(T=298.15, x=0)
H2O_l = H2O_g.h(T=298.15) - H2O_fg

#Calculate the enthalpy of a substance in liquid state at a given temperature.
def make_liquid(substance, Temp):
    
    substance_g = pm.get(f'ig.{substance}')
    substance_f = pm.get(f'mp.{substance}')
    substance_fg = substance_f.h(T=Temp, x=1) - substance_f.h(T=Temp, x=0)
    substance_l = substance_g.h(T=Temp) - substance_fg
    return substance_l[0]
```

```python
#Input/output with the user with reaction

#input("Get the enthalpy for substance: ")
temp = 298.15
input_substance = "O2"
input_substance_count = 2
input_substance2 = "CH4"
input_substance2_count = 1
output_substance = "CO2"
output_substance_count = 1
output_substance2 = "H2O"
output_substance2_count = 2

def g(sub):
    substance = pm.get(f'ig.{sub}')
    substance_enthalpy = substance.h(T=temp)
    return substance_enthalpy

test=(output_substance_count*(g(output_substance))+(output_substance2_count*g(output_substance2))-input_substance2_count*g(input_substance2)-input_substance_count*g(input_substance))[0]
print(f'{test/1000}' "KJ/mol")
```
